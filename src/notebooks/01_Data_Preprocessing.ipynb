Validate the raw dataset and inspect invalid zero measurements. Production transformations remain centralized in `src/ml_pipeline.py`.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent.parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))
from ml_pipeline import FEATURE_NAMES, MEASUREMENT_COLUMNS, load_dataset, prepare_features
DATA_PATH = PROJECT_ROOT / "src/data/diabetes.csv"
dataset = load_dataset(DATA_PATH)
dataset.head()
```

```python
print(f"Rows: {len(dataset):,}")
print(f"Duplicate rows: {dataset.duplicated().sum()}")
display(dataset.isna().sum().to_frame("missing_values"))
display((dataset == 0).sum().sort_values(ascending=False).to_frame("zero_count"))
```

```python
features, target = prepare_features(dataset)
print(f"Feature shape: {features.shape}")
print(f"Target distribution: {target.value_counts().to_dict()}")
display(features.head())
```
