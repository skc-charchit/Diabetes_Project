# 02 - Model Training

Train the production SVM pipeline with imputation, scaling, calibration, and cross-validated hyperparameter search.

```python
import json
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent.parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))
from ml_pipeline import train_model
DATA_PATH = PROJECT_ROOT / "src/data/diabetes.csv"
MODEL_PATH = PROJECT_ROOT / "src/models/diabetes_pipeline.pkl"
METRICS_PATH = PROJECT_ROOT / "src/models/training_metrics.json"
```

```python
result = train_model(DATA_PATH, MODEL_PATH)
metrics = result["metrics"]
METRICS_PATH.write_text(json.dumps(metrics, indent=2, default=str) + "\n")
print(json.dumps(metrics, indent=2, default=str))
```

The model is saved to `src/models/diabetes_pipeline.pkl`; metrics are saved to `src/models/training_metrics.json`. Use `python src/train.py` for command-line retraining.
