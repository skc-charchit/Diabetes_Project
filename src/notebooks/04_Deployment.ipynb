# 04 - Deployment

Verify the FastAPI deployment contract locally and record the production commands.

```python
import sys
from pathlib import Path

from fastapi.testclient import TestClient

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent.parent
sys.path.insert(0, str(PROJECT_ROOT))
from main import app
client = TestClient(app)
health_response = client.get("/health")
print(health_response.status_code, health_response.json())
```

```python
payload = {
    "Pregnancies": 6, "Glucose": 148, "BloodPressure": 72,
    "SkinThickness": 35, "Insulin": 125, "BMI": 33.6,
    "DiabetesPedigreeFunction": 0.627, "Age": 50,
}
response = client.post("/diabetes_prediction", json=payload)
print(response.status_code)
print(response.json())
```
